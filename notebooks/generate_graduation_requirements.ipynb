# Graduation Requirements — Dataset Documentation

**Dataset:** `graduation_requirements.csv`

**Purpose:** Defines the minimum subject-specific and total credit requirements for each graduation cohort within the synthetic Graduation Tracker student information system. Supports calculation of graduation credit deficiencies, evaluation of student progress toward graduation, and identification of students at risk of not meeting academic graduation requirements.

**Record level:** One row per graduation cohort and credit requirement category.

**Primary key:** Composite key (`cohort`, `course_subject`).

**Fields:**

- `cohort`: Expected graduation year associated with the credit requirement (2027, 2028, 2029, or 2030).
- `course_subject`: Subject classification or overall credit requirement category used to determine graduation credit requirements.
- `required_credits`: Minimum number of credits required for the specified subject or overall graduation requirement.

**Business rules:**

- Graduation requirements are defined for four graduation cohorts: 2027, 2028, 2029, and 2030.
- All graduation cohorts follow the same credit requirements.
- Each cohort contains nine requirement categories:
  - English Language Arts (`ela`): 4.0 credits.
  - Mathematics (`math`): 3.0 credits.
  - History (`hist`): 3.0 credits.
  - Science (`sci`): 3.0 credits.
  - Health (`heal`): 0.5 credits.
  - Physical Education (`pe`): 0.5 credits.
  - Art (`art`): 1.0 credit.
  - Personal Finance (`pfin`): 0.5 credits.
  - Total Credits (`total`): 25.0 credits.
- Each graduation cohort must have exactly one record for each requirement category.
- The combination of `cohort` and `course_subject` must be unique.
- All required credit values must be greater than zero.
- Subject-specific requirements total 15.5 credits.
- The overall graduation requirement is 25.0 credits, allowing 9.5 additional credits beyond the explicitly defined subject requirements.
- The `total` category represents the overall minimum credit requirement and must not be added to the subject-specific requirements when calculating graduation credit totals.
- Graduation readiness calculations must evaluate both subject-specific credit requirements and the overall 25-credit requirement.
- Credit requirements are stored as numeric values to support fractional credits.

**Data validation:**

- Verify that each (`cohort`, `course_subject`) combination is unique.
- Verify that all required credit values are greater than zero.
- Confirm that each cohort contains nine requirement categories.
- Confirm that each cohort has an overall requirement of 25.0 credits.
- Confirm that subject-specific requirements total 15.5 credits per cohort.
- Confirm that all four graduation cohorts are represented.

**Data source:** Project-defined synthetic graduation requirements created using a Python dictionary and expanded into cohort-specific records using pandas.

**Generation method:** Graduation requirement records are generated deterministically using Python and pandas. A predefined dictionary of credit requirements is applied to each graduation cohort from 2027 through 2030. The resulting DataFrame is exported as a CSV file without an index.

**Limitations:** Graduation requirements are project-defined assumptions and do not represent universal Pennsylvania statewide graduation credit requirements. Actual graduation requirements may vary by school, district, charter school, graduation cohort, and local graduation policies. The dataset defines academic credit thresholds only and does not include Keystone assessment requirements, Act 158 pathway completion, community service requirements, or other graduation eligibility criteria. Elective credits are not represented as a separate requirement category, and the difference between subject-specific requirements and the overall credit requirement is not explicitly classified.

```python
# Set random seed

rng = np.random.default_rng(42)
```

```python
# Create graduation requirements list

required = {
    'ela':4.0,
    'math':3.0,
    'hist':3.0,
    'sci':3.0,
    'heal':0.5,
    'pe':0.5,
    'art':1.0,
    'pfin':0.5,
    'total':25.0
}
```

```python
# Create graduation requirements dataframe

requirements = pd.DataFrame([
    {
        'cohort':cohort,
        'course_subject':subject,
        'required_credits':credits
    }
    for cohort in range(2027, 2031)
    for subject, credits in required.items()
])

display(requirements)
```

```python
# Validate requirements dataset

assert not requirements.duplicated(['cohort', 'course_subject']).any()

assert requirements.required_credits.gt(0).all()

print(requirements.groupby('cohort').required_credits.sum())
```

```python
# Save CSV to Google Drive

csv_path = folder_path / 'graduation_requirements.csv'

requirements.to_csv(csv_path, index = False)
```
