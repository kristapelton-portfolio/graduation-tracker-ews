### Course Catalogue — Dataset Documentation

**Dataset:** `course_catalogue.csv`

**Purpose:** Defines the courses available within the synthetic Graduation Tracker student information system.

**Record level:** One row per unique course.

**Primary key:** `course_id`

**Fields:**
- `course_id`: Unique course identifier using a subject-specific prefix.
- `course`: Course name.
- `course_subject`: Subject classification used for graduation credit calculations.
- `credit`: Credit value awarded upon successful course completion.

**Business rules:**
- Health courses award 0.5 credits.
- Physical Education courses award 0.5 credits.
- Personal Finance courses award 0.5 credits.
- All other courses award 1.0 credit.
- Course IDs must be unique.
- No required fields may be missing.

**Data source:** Synthetic course catalogue created for portfolio demonstration.

**Limitations:** Course offerings and credit values are project-defined assumptions, not statewide Pennsylvania graduation requirements.

```python
# Create course lists

ela_course = [
    'English 1',
    'English 2',
    'English 3',
    'English 4',
    'English 1 Honors',
    'English 2 Honors',
    'English 3 Honors',
    'English 4 Honors',
    'AP Language',
    'AP Literature'
]

ela_id = [f"ELA{i:05d}" for i in range(1, len(ela_course) + 1)]

math_course = [
    'Algebra 1',
    'Geometry',
    'Algebra 2',
    'Pre-Calculus',
    'Probability & Statistics',
    'Business Math',
    'Math Fundamentals',
    'Trigonometry'
]

math_id = [f"MATH{i:05d}" for i in range(1, len(math_course) + 1)]

hist_course = [
 'World History',
 'American History',
 'Political Science & Economics',
 'Psychology',
 'Sociology',
 'World History Honors',
 'American History Honors',
 'AP Government',
 'AP Psychology'
]

hist_id = [f"HIST{i:05d}" for i in range(1, len(hist_course) + 1)]

sci_course = [
    'Earth/Space Science',
    'Biology',
    'Chemistry',
    'Physics',
    'Environmental Science',
    'Marine Science',
    'Forensic Science',
    'AP Biology',
    'AP Environmental Science'
]

sci_id = [f"SCI{i:05d}" for i in range(1, len(sci_course) + 1)]

heal_course = ['Health']

heal_id = [f"HEAL{i:05d}" for i in range(1, len(heal_course) + 1)]

pe_course = [
    'Team Sports',
    'Individual Sports',
    'Weight Lifting'
]

pe_id = [f"PE{i:05d}" for i in range(1, len(pe_course) + 1)]

art_course = [
    'Visual Arts 1',
    'Visual Arts 2',
    'Digital Design 1',
    'Digital Design 2',
    'AP Visual Arts',
    'Band',
    'Music Theory',
    'Chorus',
    'AP Music Theory',
    'Theatre 1',
    'Theatre 2'
]

art_id = [f"ART{i:05d}" for i in range(1, len(art_course) + 1)]

pfin_course = ['Personal Finance']

pfin_id = [f"PFIN{i:05d}" for i in range(1, len(pfin_course) + 1)]

elec_course = [
    'Freshman Seminar',
    'Sophomore Seminar',
    'Junior Seminar',
    'Senior Seminar',
    'Computer Programming 1',
    'Computer Programming 2',
    'AP Computer Science',
    'JROTC 1',
    'JROTC 2',
    'JROTC 3',
    'JROTC 4',
    'Spanish 1',
    'Spanish 2',
    'Spanish 3',
    'Spanish 4'
]

elec_id = [f"ELEC{i:05d}" for i in range(1, len(elec_course) + 1)]
```

```python
# Create course catalogue

# Connect subject code to course list

course_subject = {
    'math': (math_course, math_id),
    'ela': (ela_course, ela_id),
    'hist': (hist_course, hist_id),
    'sci': (sci_course, sci_id),
    'heal': (heal_course, heal_id),
    'pe': (pe_course, pe_id),
    'art': (art_course, art_id),
    'pfin': (pfin_course, pfin_id),
    'elec': (elec_course, elec_id),
}

# Add credits to courses

half_credit = ['heal', 'pe', 'pfin']

course_data = []

for subject, (course_list, course_ids) in course_subject.items():
  if subject in half_credit:
    credit = 0.5
  else:
    credit = 1
  for course_id, course in zip(course_ids, course_list):
    course_data.append({
        'course_id': course_id,
        'course': course,
        'course_subject': subject,
        'credit': credit
    })

course_catalogue = pd.DataFrame(course_data)

display(course_catalogue)
```

```python
# Verify course catague creation

# Number of Courses and columns
print(len(course_catalogue))
print(course_catalogue.columns.tolist())

# Data Types
print(course_catalogue.dtypes)

# Duplicate Course IDs
print(course_catalogue['course_id'].duplicated().sum())

# Duplicate Course Names
print(course_catalogue['course'].duplicated().sum())

# Missing Values
print(course_catalogue.isnull().sum())

# Verify half credit courses
course_catalogue['expected_credit'] = (
    course_catalogue['course_subject'].apply(lambda x: 0.5 if x in half_credit else 1.0)
)

# Incorrect credit assignments
credit_errors = course_catalogue['credit'] != course_catalogue['expected_credit']
print(len(credit_errors))
```

```python
# Drop temporary column

course_catalogue.drop(
    columns = ['expected_credit'],
    inplace = True
)
```

```python
# Verify course catalogue creation

# Confirm 67 courses
assert len(course_catalogue) == 67

# Confirm all IDs are unique
assert course_catalogue['course_id'].is_unique

# Confirm course names are unique
assert course_catalogue['course'].is_unique

# Confirm no missing values
assert not course_catalogue.isnull().any().any()

# Confirm exactly 5 half credit courses
assert (course_catalogue['credit'] == 0.5).sum() == 5
```
